```python
import ee 
import geesat 
from geesat import geogee, geosen
import geopandas as gpd
# ee.Authenticate()
# ee.Initialize()
geesat.authenticate_gee()
import geemap
import math
```

```python
# Example of area of interest (AOI) in Germany
bbox = [9.84375   , 47.5172007 , 10.1953125 , 47.75409798]
# Create an Earth Engine geometry object for the AOI
aoi = ee.Geometry.Rectangle(bbox)
```

```python
water_occurrence = geosen.generate_water_occurrence(roi=aoi, start_date='2024-01-01', end_date='2025-12-31',
orbit_pass='Descending',
polarization='VH'
)
print(water_occurrence.bandNames().getInfo())
```
