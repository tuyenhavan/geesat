```python
import ee 
import geesat 
from geesat import geogee, geosen
import geopandas as gpd
# ee.Authenticate()
# ee.Initialize()
geesat.authenticate_gee()
import geemap
import math
```

```python
# Example of area of interest (AOI) in Germany
bbox = [9.84375   , 47.5172007 , 10.1953125 , 47.75409798]
# Create an Earth Engine geometry object for the AOI
aoi = ee.Geometry.Rectangle(bbox)
```

```python
sen1col = geosen.prepare_sentinel1_collection(
    roi=aoi,
    orbit_pass="BOTH",
    start_date='2025-01-01',
    end_date='2025-12-31',
)

sen1col = sen1col.filter(ee.Filter.calendarRange(6, 10, 'month'))  # Filter to include only images from June to October
water_occurrence = geosen.generate_water_occurrence(roi=aoi, start_date='2025-01-01', end_date='2025-12-31',
orbit_pass='both',
polarization='VV'
)
print(water_occurrence.bandNames().getInfo())
```
